```python
# imports
import os
import json
from dotenv import load_dotenv
from IPython.display import Markdown, display, update_display
from scraper import fetch_website_links, fetch_website_contents
from openai import OpenAI
```

```python
# constants

MODEL_GPT = 'gpt-4o'
MODEL_LLAMA = 'llama3.2'
load_dotenv(override=True)
api_key = os.getenv('OPENAI_API_KEY')
```

```python
#generator
def get_next_book(book_list):
    for dict in book_list:
        yield f"book name {dict['name']} by the author {dict['author']}"
```

```python
# here is the question; type over this to ask something new

system_prompt_book_list = """
you are a book specialist providing the list of dictionary in the format
the output format should look like this:
[
    {"name": "To Kill a Mockingbird", "author": "Harper Lee"},
    {"name": "1984", "author": "George Orwell"}
]
this is just the example of the format.
"""

user_prompt_book_list = f"""
give me the top 7 all time best selling books.
"""
```

```python
message_book_list = [
    {"role":"system","content":system_prompt_book_list},
    {"role":"system","content":user_prompt_book_list}
]
message_book_list
```

```python
llm_model = OpenAI(api_key=api_key)

def get_book_list():
    response = llm_model.chat.completions.create(
        model=MODEL_GPT,
        messages=message_book_list
    )
    result = response.choices[0].message.content
    result = json.loads(result)
    return result
```

```python
system_prompt_book_summary = """
you are a goofy book specialist providing summary in less than 100 words.
you provide the output in markdown format which looks cool and beautiful.
"""

user_prompt_book_summary = f"""
give me the detailed summary of the books and author provided in less than 100 words.
summary should include the overview of book and information about the author.
"""
```

```python
message_book_summary = [
    {"role":"system", "content": system_prompt_book_summary},
    {"role":"user", "content":  user_prompt_book_summary}
]
message_book_summary
```

```python
def book_summary(model,api_key):
    book_list = get_book_list()
    for book in book_list:
        message_book_summary[1]["content"] += f"Book name : {book['name']} by author {book['author']} \n"

    llm_model = OpenAI(api_key=api_key)
    stream = llm_model.chat.completions.create(
        model=model,
        messages=message_book_summary,
        stream=True
    )
    response = ""
    display_handle = display(Markdown(""), display_id=True)
    for chunk in stream:
        response += chunk.choices[0].delta.content or ''
        update_display(Markdown(response), display_id=display_handle.display_id)

book_summary('gpt-5.1',api_key=api_key)
```
